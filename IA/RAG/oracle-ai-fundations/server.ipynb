```python
# sqlite3: modulo de base de datos liviana local
import sqlite3

# typing: libreria nativa de python para establecer tipos complejos
# List: una variable o una salida espera ser un lista
# Optional: nos facilita el trabajo al definir un None implicito
from typing import List, Optional

# pydantic: libreria externa para validación de datos y gestión de configuraciones
from pydantic import BaseModel, Field

# fastmpc: nos permite crear un servidor python para nuestros mcp
from fastmcp import FastMCP


# Inicializar aplicación y asignarle un nombre
my_mcp=FastMCP("mcp:sqlite-user-crud")

# Determinamos el nombre de la base de datos
db_name="mcp-test-db"

# Crear funcion que inicia base de datos si no existe
def init_db():

    """create database if not exists, when server initialize"""

    # iniciar contexto
    with sqlite3.connect(db_name) as conn:

        # Connect devuelve el objeto de conexión y con éste se crea el intermediario(cursor)
        cursor = conn.cursor()

        # Cursor tiene métodos
        cursor.execute("""
                        CREATE TABLE IF NOT EXISTS usuarios(
                            id INTEGER PRIMARY KEY AUTOINCREMENT,
                            nombre TEXT NOT NULL,
                            email TEXT NOT NULL,
                            telefono TEXT
                        )
                       """)
        conn.commit()

init_db()

# Validacion de schemas(estructura de datos)

# Para crear y ver usuarios
class UsuarioCreate(BaseModel):
    nombre:str = Field(..., description="Nombre completo del usuario")
    email:str = Field(..., description="Correo electronico del usuario")
    telefono:Optional[str] = Field(None, description="Celular del usuario")

class UsuarioUpdate(BaseModel):
    nombre: Option[str] = Field(None, description="Nombre completo del usuario")
    email: Option[str] = Field(None, description="Correo electronico del usuario")
    telefono: Option[str] = Field(None, description="Celular del usuario")

# Creacion de Herramientas para el MCP
@mcp.tool()
```
